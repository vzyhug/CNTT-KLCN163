# Table IV models: metrics by noise intensity (QTDB, Kaggle)

Evaluate the **13 saved models in Table IV of `paper-eng.tex`**, without training.
Each metric has its own table: model rows and four noise-intensity columns.

**Run All on Kaggle:**
1. Add `common_qtdb_mixed_bwem_ma_3600_ntype1.npz`.
2. Add either the previous evaluation output (13 `*_predictions.npz` plus
   `evaluation_metadata.json`), or the saved checkpoints: Proposed `.pth` and
   `comparative_checkpoints_bundle.zip` (or the corresponding loose H5/PTH files).
3. With checkpoints only, enable GPU. The existing Table IV inference pipelines
   are embedded below, including architecture source; no training or git clone.

`MODE="auto"` uses validated predictions if all 13 are available; otherwise it
runs the saved-checkpoint evaluation. `MODE="cached"` needs no GPU/TensorFlow.
`MODE="inference"` executes the checkpoint pipeline; its existing verified caches
for Proposed/TFCDiff can still be reused.

Outputs: seven metric tables (**SSD, RMSE, MAD, PRD, cosine similarity, output SNR,
SNR improvement**) in CSV, HTML, and LaTeX; numeric mean/SD tables, sample-level
metrics, bin counts, an overall summary, and a reproducibility receipt.

## Context & methods

Bins use the dataset's **`q_test` peak-to-peak noise scaling coefficient**, not
input SNR in dB: `[0.2,0.6)`, `[0.6,1.0)`, `[1.0,1.5)`, `[1.5,2.0]`.
Boundary values 0.6, 1.0, and 1.5 belong to the bin on their right; 2.0 belongs
to the last bin. This is a stratification of the existing test pairs, not a new
noise generation experiment; the same pairs and bin membership apply to every model.

Compute metrics **per segment first**, then report arithmetic mean ± sample SD
within each bin. PRD is Table IV's mean-centered-reference definition:
`100 * sqrt(SSD / sum((clean - mean(clean))**2))`.
MAD means **maximum** absolute difference. Cosine similarity is uncentered.
SSD/RMSE/MAD/PRD: lower is better; CosSim/SNR_out/DeltaSNR: higher is better.

### Key assumptions
Preserve Table IV evaluation lengths: 3600 points except DF--FCN-DAE,
CBAM--Attention-Skip, and CBAM--ECA-Skip at **3584** (marked `*`).
Those three rows are a separate comparison group; do not rank them jointly with
3600-point rows. No output renormalization is performed during metric calculation.
The model pipelines, seed (2024), and checkpoints match the original evaluation.
SD describes variation across segments, not uncertainty across independent runs.
No significance claim follows from the means alone. Latency is omitted because
the original pipelines use different timing scopes.

### 2. Read the common test set and audit noise bins

```python
with np.load(data_path, allow_pickle=False) as z:
    clean = np.ascontiguousarray(z['clean_test'], dtype=np.float32)
    noisy = np.ascontiguousarray(z['noisy_test'], dtype=np.float32)
    sample_id = z['sample_id_test'].astype(np.int64)
    record_id = z['record_id_test'].astype(str)
    q_test = z['q_test'].astype(np.float64)
    train_val = set()
    for key in ('record_id_train', 'record_id_val'):
        if key in z.files: train_val.update(z[key].astype(str))
assert clean.shape == noisy.shape == (5012, 1, 3600)
assert len(sample_id) == len(np.unique(sample_id)) == len(record_id) == len(q_test) == 5012
assert not set(record_id).intersection(train_val), 'Held-out records overlap train/validation.'
assert np.isfinite(clean).all() and np.isfinite(noisy).all()
bin_index = intensity_bins(q_test)
counts = pd.DataFrame({
    'Noise intensity q': BIN_LABELS,
    'Interval': ['[0.2,0.6)', '[0.6,1.0)', '[1.0,1.5)', '[1.5,2.0]'],
    'N': np.bincount(bin_index, minlength=4),
    'Records': [len(np.unique(record_id[bin_index == i])) for i in range(4)],
})
assert counts.N.gt(0).all() and counts.N.sum() == len(sample_id)
counts.to_csv(TABLE_OUT / 'noise_bin_counts.csv', index=False)
display(counts)
```

### 3. Validate saved predictions or select checkpoint inference

Every archive must have the exact dataset hash, unique sample IDs, expected
output length, and finite predictions. IDs are explicitly aligned to the common
test order. The checkpoint identity is checked against the archive or the
adjacent evaluation receipt. Ambiguous matches require `PREDICTION_DIR`.

```python
def prediction_array(saved, length):
    p = np.asarray(saved['prediction'], dtype=np.float32)
    if p.ndim == 3 and p.shape[1] == 1: p = p[:, 0, :]
    elif p.ndim == 3 and p.shape[2] == 1: p = p[:, :, 0]
    if p.shape != (len(sample_id), length) or not np.isfinite(p).all():
        raise ValueError(f'Invalid prediction shape/values: {p.shape}, expected {(len(sample_id),length)}')
    return p

def validate_archive(path, spec, return_prediction=False):
    name, label, filename, length, weight_rel = spec
    expected_weight = EXPECTED_PROPOSED_SHA256 if name == 'Proposed' else WEIGHT_HASHES[weight_rel]
    with np.load(path, allow_pickle=False) as z:
        if str(z['data_sha256'].item()) != data_hash: raise ValueError('Dataset hash mismatch')
        ids = np.asarray(z['sample_id'], dtype=np.int64)
        if ids.shape != sample_id.shape or len(np.unique(ids)) != len(ids) or set(ids) != set(sample_id):
            raise ValueError('Missing, duplicate, or unrelated sample IDs')
        pred = prediction_array(z, length)
        if 'checkpoint_sha256' in z.files:
            actual_weight = str(z['checkpoint_sha256'].item())
        else:
            receipt = json.loads((path.parent / 'evaluation_metadata.json').read_text(encoding='utf-8'))
            if receipt['data_sha256'] != data_hash: raise ValueError('Receipt dataset mismatch')
            actual_weight = receipt.get('comparative_weight_sha256', {}).get(weight_rel)
        if actual_weight != expected_weight: raise ValueError(f'Checkpoint identity mismatch: {name}')
        if return_prediction:
            order = {int(v): i for i,v in enumerate(ids)}
            return pred[[order[int(v)] for v in sample_id]]
    return True

prediction_paths, cache_problems = {}, {}
for spec in MODEL_SPECS:
    roots = [Path(PREDICTION_DIR)] if PREDICTION_DIR else SEARCH_ROOTS
    candidates = find_files(spec[2], roots)
    valid, problems = [], []
    for path in candidates:
        try:
            validate_archive(path, spec)
            valid.append(path)
        except Exception as exc:
            problems.append(f'{path}: {exc}')
    if len(valid) > 1:
        raise RuntimeError(f'Multiple validated caches for {spec[0]}; set PREDICTION_DIR: {valid}')
    if valid: prediction_paths[spec[0]] = valid[0]
    else: cache_problems[spec[0]] = problems or ['No prediction archive found']

NEED_INFERENCE = MODE == 'inference' or len(prediction_paths) != len(MODEL_SPECS)
if MODE == 'cached' and NEED_INFERENCE:
    raise RuntimeError('MODE=cached needs all 13 valid predictions and provenance: ' + json.dumps(cache_problems, indent=2))
print('Validated cached models:', len(prediction_paths), '/', len(MODEL_SPECS))
print('Run saved-checkpoint inference:', NEED_INFERENCE)
```

### 6. Calculate per-segment metrics, preserving each model’s evaluation length

```python
METRICS = {
    'SSD': ('SSD (lower is better)', 3),
    'RMSE': ('RMSE (lower is better)', 4),
    'MAD': ('MAD: maximum absolute difference (lower is better)', 4),
    'PRD': ('PRD, mean-centered reference (%) (lower is better)', 3),
    'CosSim': ('Cosine similarity (higher is better)', 4),
    'SNR_out': ('Output SNR (dB) (higher is better)', 3),
    'DeltaSNR': ('SNR improvement (dB) (higher is better)', 3),
}

def segment_metrics(reference, observed, prediction):
    x, y, p = [np.asarray(v, dtype=np.float64) for v in (reference, observed, prediction)]
    assert x.ndim == 2 and x.shape == y.shape == p.shape
    e = p-x
    ssd = np.sum(e*e, axis=1)
    energy = np.sum(x*x, axis=1)
    centered_energy = np.sum((x-x.mean(axis=1, keepdims=True))**2, axis=1)
    pred_energy = np.sum(p*p, axis=1)
    noise_energy = np.sum((y-x)**2, axis=1)
    if np.any(energy <= 0) or np.any(centered_energy <= 0) or np.any(pred_energy <= 0):
        raise ValueError('Zero-energy reference or prediction: undefined PRD/cosine, not silently clamped.')
    with np.errstate(divide='ignore'):
        snr_in = 10*np.log10(energy/noise_energy)
        snr_out = 10*np.log10(energy/ssd)
    return pd.DataFrame({
        'SSD': ssd, 'RMSE': np.sqrt(ssd/x.shape[1]),
        'MAD': np.max(np.abs(e), axis=1),
        'PRD': 100*np.sqrt(ssd/centered_energy),
        'CosSim': np.sum(x*p, axis=1)/np.sqrt(energy*pred_energy),
        'SNR_in': snr_in, 'SNR_out': snr_out, 'DeltaSNR': snr_out-snr_in,
    })

frames, archive_receipts = [], []
for spec in MODEL_SPECS:
    name, label, filename, length, weight_rel = spec
    path = prediction_paths[name]
    pred = validate_archive(path, spec, return_prediction=True)
    batches = []
    for start in range(0, len(sample_id), 128):
        stop = min(start+128, len(sample_id))
        batches.append(segment_metrics(clean[start:stop,0,:length], noisy[start:stop,0,:length], pred[start:stop]))
    frame = pd.concat(batches, ignore_index=True)
    frame.insert(0, 'model', name)
    frame['metric_length'] = length
    frame['sample_id'], frame['record_id'], frame['q'] = sample_id, record_id, q_test
    frame['noise_bin'] = pd.Categorical(np.array(BIN_LABELS)[bin_index], categories=BIN_LABELS, ordered=True)
    if not np.isfinite(frame[list(METRICS)+['SNR_in']].to_numpy()).all():
        raise ValueError(f'{name}: undefined/infinite metric; inspect segment-level inputs before reporting means.')
    frames.append(frame)
    archive_receipts.append({'model':name, 'file':str(path), 'sha256':sha256_file(path), 'metric_length':length})
    del pred
detail = pd.concat(frames, ignore_index=True)
assert len(detail) == 13*5012
assert not detail.duplicated(['model','sample_id']).any()
assert detail.groupby(['model','noise_bin'], observed=True).size().unstack().reindex(MODEL_ORDER).eq(counts.N.to_numpy()).all().all()
detail.to_csv(TABLE_OUT/'per_segment_metrics.csv.gz', index=False, compression='gzip')
summary = detail.groupby('model', sort=False)[list(METRICS)+['SNR_in']].mean().reindex(MODEL_ORDER)
summary.to_csv(TABLE_OUT/'overall_metrics.csv')
print('Computed', len(detail), 'model/segment rows; all models share exactly the same bin counts.')
```

## Results

Each table below has the requested four intensity columns. Cells are
**mean ± sample SD**, calculated from the actual segments in that bin.
`*` marks a 3584-point model; all other rows use 3600 points.
The CSV files ending in `_mean.csv` contain plain numeric means for paper tables.

```python
def metric_tables(metric):
    grouped = detail.groupby(['model','noise_bin'], observed=True)[metric]
    mean = grouped.mean().unstack().reindex(index=MODEL_ORDER, columns=BIN_LABELS)
    sd = grouped.std(ddof=1).unstack().reindex(index=MODEL_ORDER, columns=BIN_LABELS)
    n = grouped.count().unstack().reindex(index=MODEL_ORDER, columns=BIN_LABELS)
    reconstructed = (mean*n).sum(axis=1)/n.sum(axis=1)
    np.testing.assert_allclose(reconstructed, summary.loc[MODEL_ORDER,metric], rtol=1e-12, atol=1e-12)
    for table in (mean,sd,n):
        table.index = [MODEL_LABELS[v] for v in MODEL_ORDER]
        table.index.name = 'Model'
        table.columns.name = 'Noise intensity q'
    decimals = METRICS[metric][1]
    formatted = pd.DataFrame(index=mean.index, columns=mean.columns)
    for col in BIN_LABELS:
        formatted[col] = [f'{m:.{decimals}f} ± {s:.{decimals}f}' for m,s in zip(mean[col],sd[col])]
    return mean, sd, n, formatted

exported_tables = {}
def show_metric(metric):
    mean, sd, n, formatted = metric_tables(metric)
    for suffix, table in [('mean',mean),('sd',sd),('n',n),('mean_sd',formatted)]:
        table.to_csv(TABLE_OUT/f'{metric}_{suffix}.csv')
    caption = METRICS[metric][0] + '; mean ± sample SD. *: 3584 points; other rows: 3600 points.'
    styler = formatted.style.set_caption(caption).set_properties(**{'text-align':'right','white-space':'nowrap'})
    styler = styler.set_table_styles([
        {'selector':'caption','props':[('font-size','15px'),('font-weight','bold'),('text-align','left'),('padding','8px')]},
        {'selector':'th','props':[('text-align','left'),('padding','6px')]},
        {'selector':'td','props':[('padding','6px')]},
    ])
    html = '<meta charset="utf-8">'+styler.to_html()
    (TABLE_OUT/f'{metric}.html').write_text(html, encoding='utf-8')
    # Escape text safely, then use LaTeX's own math symbol for ±.
    latex = formatted.to_latex(escape=True, column_format='lrrrr', caption=caption,
                              label=f'tab:intensity_{metric.lower()}').replace('±', r'$\pm$')
    (TABLE_OUT/f'{metric}.tex').write_text(latex, encoding='utf-8')
    exported_tables[metric] = {'html':html,'latex':latex}
    display(styler)
```

## Fine-grained noise-intensity curves — no additional inference

Plot existing per-segment metrics at finer q intervals. All 13 models are
shown in separate comparison panels; the three 3584-point models remain
separate from the 3600-point comparisons. Points represent means within
intervals, not repeated corruption of the same clean ECG.

Set `CURVE_RECORD = "sel123"` to restrict the plot to that held-out record,
or leave `None` to use all 14 test records. The 0.10 interval width gives
18 points instead of four. Smaller widths are allowed, but may leave too
few samples per point, especially for a single record. Counts and SD are
exported; SD is segment variability, not a confidence interval. Lines merely
connect descriptive estimates; they do not imply continuous measurements.

For a **controlled sweep**, choose clean segments and noise waveforms before
inspecting model outcomes, hold them fixed, and vary only q. From an existing
pair at q_i > 0, its fixed noise waveform can be defined as
`n_i = (noisy_i - clean_i) / q_i`; then construct
`noisy_i(q) = clean_i + q*n_i`. This requires fresh model inference but no
retraining. A single record is an illustrative case study, not evidence of
generalization across patients. Select diffusion seeds and normalization rules
consistently across levels and report the sampling policy.

```python
import matplotlib.pyplot as plt
CURVE_RECORD = None  # e.g. "sel123"; None = all held-out records
CURVE_Q_WIDTH = 0.10  # 0.10 -> 18 intervals; 0.05 -> 36
CURVE_MIN_SAMPLES = 10
curve_dir = TABLE_OUT / 'fine_curves'
curve_dir.mkdir(exist_ok=True)
curve_data = detail.copy()
if CURVE_RECORD is not None:
    curve_data = curve_data[curve_data.record_id == CURVE_RECORD].copy()
    if curve_data.empty: raise ValueError(f'Unknown test record: {CURVE_RECORD}')
width_cents = int(round(CURVE_Q_WIDTH*100))
if width_cents < 1 or not np.isclose(width_cents/100, CURVE_Q_WIDTH) or 180 % width_cents:
    raise ValueError('Use a width in 0.01 increments that divides the 0.2–2.0 range.')
q_cents = np.rint(curve_data.q.to_numpy()*100).astype(int)
if not np.allclose(q_cents/100, curve_data.q) or np.any((q_cents<20)|(q_cents>200)):
    raise ValueError('Unexpected q grid or range')
curve_data['q_bin'] = np.minimum((q_cents-20)//width_cents, 180//width_cents-1)
curve_data['q_center'] = 0.2+(curve_data.q_bin+0.5)*CURVE_Q_WIDTH
curve_stats = curve_data.groupby(['model','metric_length','q_bin','q_center'], observed=True)[list(METRICS)].agg(['mean','std','count'])
curve_stats.columns = ['_'.join(c) for c in curve_stats.columns]
curve_stats = curve_stats.reset_index()
scope = CURVE_RECORD or 'all_records'
curve_stats.to_csv(curve_dir/f'fine_curve_metrics_{scope}.csv', index=False)
curve_counts = curve_data[curve_data.model=='Proposed'].groupby(['q_bin','q_center'],observed=True).size()
display(curve_counts.rename('N segments per model').to_frame())
print(f'{scope}: {len(curve_counts)} intervals; N/point {curve_counts.min()}–{curve_counts.max()}')
print('Existing predictions only: no model inference was rerun.')
```

```python
CURVE_GROUPS = [
    ('Diffusion models | 3600 points', ['Proposed','TFCDiff','DeScoD']),
    ('Transformer comparisons | 3600 points', ['Proposed','TCDAE','CBAM_Transformer_DAE']),
    ('DeepFilter comparisons | 3600 points', ['Proposed','DeepFilter_DRNN','DeepFilter_Vanilla_L','DeepFilter_Vanilla_NL','DeepFilter_Multibranch_LANL','DeepFilter_Multibranch_LANLD']),
    ('Separate evaluation group | 3584 points', ['DeepFilter_FCN_DAE','CBAM_Attention_Skip_DAE','CBAM_ECA_Skip_DAE']),
]
other_models = [v for v in MODEL_ORDER if v != 'Proposed']
palette = ['#2166ac','#b27d12','#c65e28','#71832f','#aa517e']
curve_colors = {name:palette[i%5] for i,name in enumerate(other_models)}
curve_styles = {name:['-','--',':','-.'][i//5] for i,name in enumerate(other_models)}
curve_colors['Proposed'], curve_styles['Proposed'] = '#222222', '-'

def plot_noise_curve(metric):
    fig, axes = plt.subplots(2,2,figsize=(15,10),sharex=True,sharey=True)
    for ax,(title,names) in zip(axes.flat,CURVE_GROUPS):
        for name in names:
            g = curve_stats[curve_stats.model==name].sort_values('q_center')
            g = g.set_index('q_bin').reindex(range(180//width_cents))
            x = 0.2+(np.arange(len(g))+0.5)*CURVE_Q_WIDTH
            values = g[f'{metric}_mean'].where(g[f'{metric}_count']>=CURVE_MIN_SAMPLES)
            ax.plot(x,values,label=MODEL_LABELS[name].replace(' *',''),
                    color=curve_colors[name],linestyle=curve_styles[name],
                    marker='o' if name=='Proposed' else None,markersize=3,linewidth=1.7)
        ax.set_title(title,fontsize=12,loc='left')
        ax.set_xlabel('Noise intensity q (peak-to-peak scaling)',fontsize=10)
        ax.set_ylabel(METRICS[metric][0],fontsize=10)
        ax.set_xlim(.2,2.0)
        ax.set_xticks(np.arange(.2,2.01,.2))
        if metric=='CosSim': ax.set_ylim(0,1)
        else: ax.set_ylim(0, float(curve_stats[f'{metric}_mean'].max())*1.08)
        ax.grid(alpha=.2)
        ax.spines[['top','right']].set_visible(False)
        ax.legend(fontsize=9,loc='best',frameon=False)
    fig.suptitle(f'{metric} versus noise intensity | {scope}',fontsize=16,y=.99)
    fig.text(.5,.015,f'Mean per q interval (width {CURVE_Q_WIDTH:.2f}); different existing segments across intervals. '
             f'Points with N < {CURVE_MIN_SAMPLES} omitted. 3584-point group evaluated separately.',
             ha='center',fontsize=10)
    fig.tight_layout(rect=(0,.045,1,.96))
    for ext in ('png','pdf'):
        fig.savefig(curve_dir/f'{metric}_vs_q_{scope}.{ext}',dpi=180,bbox_inches='tight')
    plt.show()
    plt.close(fig)
```
